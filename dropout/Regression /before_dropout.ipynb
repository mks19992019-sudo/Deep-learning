```python
import tensorflow as tf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense , Dropout
from tensorflow.keras.optimizers import Adam
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score , mean_squared_error ,r2_score
```

```python
X_train = np.linspace(-1,1,20)
X_train
```

```python

Y_train = np.random.uniform(-1,1,20)
Y_train
```

```python
X_test=np.linspace(-1,1,20)
```

```python
Y_test =  np.random.uniform(-1,1,20)
Y_test
```

```python
plt.scatter(x=X_train,y= Y_train,c='black',label=True)
plt.scatter(x=X_test,y= Y_test,c='red',label=True)
plt.legend()

```

```python
model = Sequential()
model.add(Dense(128,input_dim=1,activation='relu'))
model.add(Dense(128,activation='relu'))
model.add(Dense(1,activation='linear'))

adam = Adam(learning_rate=0.01)
model.compile(loss='mse',optimizer=adam)
history = model.fit(X_train,Y_train,epochs=500,validation_data=(X_test,Y_test),verbose=1)
```

```python
model.evaluate(X_train,Y_train)
```

```python
model.evaluate(X_test,Y_test)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
r2_score(Y_test,y_pred)
```

```python
plt.scatter(x=X_train,y= Y_train,c='black',label=True) # balck training
plt.scatter(x=X_test,y= Y_test,c='red',label=True) # testing data
plt.plot(X_test,y_pred)
plt.ylim((-9.5,9.5))


```

```python
plt.plot(history.history['loss'],label='train_data loss')
plt.plot(history.history['val_loss'],label='test_data loss')
```
