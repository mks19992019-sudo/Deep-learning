```python
import tensorflow
from tensorflow.keras.layers import Dropout, Dense
from tensorflow.keras.models import Sequential
import  pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.datasets import make_classification


```

```python
X , Y = make_classification(n_samples=200,n_features=2,n_informative=2,n_redundant=0,n_classes=2,random_state=42)
```

```python
plt.scatter(X[:,0],X[:,1],c=Y)
```

```python
model = Sequential()
model.add(Dense(128,activation='relu',input_dim=2))
model.add(Dense(128,activation='relu'))
model.add(Dense(1,activation='sigmoid'))

adam = Adam(learning_rate=0.01)
model.compile(loss='binary_crossentropy',optimizer=adam,metrics=['accuracy'])
history = model.fit(X,Y ,epochs=500,validation_split=.2,verbose=1)
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
class ModelWrapper:
    def __init__(self, model):
        self.model = model

    def predict(self, X):
        predictions = self.model.predict(X, verbose=0)
        return (predictions > 0.5).astype(int).ravel()

clf = ModelWrapper(model)

plot_decision_regions(X, Y.ravel(), clf=clf)
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.ylim(-3,3)
```

```python
model2 = Sequential()
model2.add(Dense(128,activation='relu',input_dim=2))
model2.add(Dropout(0.5))
model2.add(Dense(128,activation='relu'))
model2.add(Dropout(0.5))
model2.add(Dense(1,activation='sigmoid'))

adam = Adam(learning_rate=0.01)
model2.compile(loss='binary_crossentropy',optimizer=adam,metrics=['accuracy'])
history2 = model2.fit(X,Y ,epochs=500,validation_split=.2,verbose=1)
```

```python
class ModelWrapper:
    def __init__(self, model):
        self.model = model

    def predict(self, X):
        predictions = self.model.predict(X, verbose=0)
        return (predictions > 0.5).astype(int).ravel()

clf = ModelWrapper(model2)

plot_decision_regions(X, Y.ravel(), clf=clf)
```

```python
plt.plot(history2.history['loss'])
plt.plot(history2.history['val_loss'])
plt.ylim(-3,3)
```

# Effect of p in dropout layer
in short if p is very less than .2 than it is overfitting and if p is very high than it is underfitting. so the best value of p is between .2 and .5.
