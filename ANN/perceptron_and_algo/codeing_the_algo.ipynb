```python
def step(z):
    return 1 if z>0 else 0
```

```python
def perceptron(X,Y):
    X= np.insert(X,[0],1,axis=1)
    weight = np.ones(X.shape[1])
    lr = 0.1
    for i in range (1000):
        j = np.random.randint(0,100)
        y_prd = step(np.dot(X[j],weight))              #([1,-0.72,0.02],[1,1,1])
        weight = weight + lr*(Y[j]-y_prd)*X[j]

    return weight[0],weight[1:]      # weight[0] => w0 or b , weight[1:] => w1,w2

        
```

```python
intercept , coef =perceptron(X,Y)
```

```python
m = -(coef[0]/coef[1])
b = -(intercept/coef[1])
print(m)
print(b)
```

```python
Y_new
```

```python
def perceptron(X_new,Y_new):
    w1=w2=b=1
    lr = 0.1
    for i in range (1000):
        #epoch 
        for j in range(X_new.shape[0]):
            z = w1*X_new[j][0] + w2*X_new[j][1] + b

            if Y_new[j]*z <= 0 :
                w1 = w1 + lr*Y_new[j]*X_new[j][0]
                w2 = w2 + lr*Y_new[j]*X_new[j][1]
                b = b + lr*Y_new[j] 

    return w1 , w2,b



        
```

```python
w1 ,w2,b = perceptron(X_new,Y_new)
```
