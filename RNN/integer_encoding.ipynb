```python
data = ['go india',
        'india india',
        'hip hip hurray',
        'jeetega bhai jeetega india jeetega',
        'bharat mata ki jai',
        'kohli kohli',
        'sachin sachin',
        'Dhoni dhoni',
        'modi ji ki jai',
        'inquilab zindabad']
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer(oov_token='<nothing>')
```

```python
tokenizer.fit_on_texts(data)
```

```python
tokenizer.word_index
```

```python
tokenizer.word_counts
```

```python
sequences =tokenizer.texts_to_sequences(data)
sequences
```

```python
from tensorflow.keras.utils import pad_sequences
```

```python
sequences = pad_sequences(sequences,padding='post') # u can aslo use pre
```

```python
sequences
```

```python
from tensorflow.keras.datasets import imdb
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Flatten
```

```python
(X_train , Y_train) , (X_test,Y_test )= imdb.load_data() 
X_train.shape
```

```python
print(X_train[0])
```

```python
X_train = pad_sequences(X_train,padding='post')
X_test = pad_sequences(X_test,padding='post')
X_train.shape
```

```python
X_train[0]
```

```python
model = Sequential()

# 2494 is temp and 1 is colum 
model.add(SimpleRNN(32,activation='relu',input_shape =(2494,1),return_sequences=False ))
model.add(Dense(1,activation='sigmoid'))
model.summary()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
model.fit(X_train,Y_train,validation_data=(X_test,Y_test),epochs=2)
```

## Embbeding layer on imdb data

```python
(X_train2 , Y_train2) , (X_test2,Y_test2 )= imdb.load_data(num_words=10000) 
print(X_train2.shape)
print(X_test2.shape)
print(Y_train2.shape)
print(Y_test2)
print(Y_test2.shape)
```

```python
X_train2=pad_sequences(X_train2,padding='post',maxlen=50)
X_test2 = pad_sequences(X_test2,padding='post',maxlen=50)
```

```python
X_train2.shape
```

```python
model5 = Sequential()
model5.add(Embedding(input_dim=10000,output_dim=2))
model5.add(SimpleRNN(32,return_sequences=False))
model5.add(Dense(1,activation='sigmoid'))
model5.compile(loss='binary_crossentropy',metrics=['accuracy'],optimizer='Adam')


model5.build(input_shape=(None,50)) # sequnce of token after pading 
model5.summary()
```

```python
model5.fit(X_train2,Y_train2,validation_data=(X_test2,Y_test2),epochs=5)
```
